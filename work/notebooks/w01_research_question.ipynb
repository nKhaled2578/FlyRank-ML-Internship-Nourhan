## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the dataset
# (Note: I'll make sure the file name matches my repo before running)
df = pd.read_csv('content_refresh_anonymized.csv')

# Getting a quick look at how scattered the impressions are
total_pages = len(df)
max_imp = df['impressions_90d'].max()
median_imp = df['impressions_90d'].median()

print(f"Total pages we are dealing with: {total_pages}")
print(f"Highest impressions (90 days) on a single page: {max_imp}")
print(f"Median impressions (90 days): {median_imp}")
print("\nTakeaway: The gap between the median and the top page is huge. We can't analyze this manually, which is why grouping them into clusters makes the most sense.")
```
